# Task: Covid-19 Data Analysis
### This notebook is used to understand the comprehension of Data Analysis techniques using Pandas library.

### Data Source: 
https://github.com/CSSEGISandData/COVID-19/tree/master/csse_covid_19_data/csse_covid_19_daily_reports

### File naming convention

MM-DD-YYYY.csv in UTC.

### Field description

- Province_State: China - province name; US/Canada/Australia/ - city name, state/province name; Others - name of the event (e.g., "Diamond Princess" cruise ship); other countries - blank.

- Country_Region: country/region name conforming to WHO (will be updated).

- Last_Update: MM/DD/YYYY HH:mm (24 hour format, in UTC).

- Confirmed: the number of confirmed cases. For Hubei Province: from Feb 13 (GMT +8), we report both clinically diagnosed and lab-confirmed cases. For lab-confirmed cases only (Before Feb 17), please refer to who_covid_19_situation_reports. For Italy, diagnosis standard might be changed since Feb 27 to "slow the growth of new case numbers." (Source)

- Deaths: the number of deaths.

- Recovered: the number of recovered cases.

### Question 5

#### Plot Country-wise Total deaths, confirmed, recovered and active casaes where total deaths have exceeded 50,000

```python
import matplotlib.pyplot as plt
```

```python
top_countries = (
    df.groupby('Country_Region')[['Confirmed', 'Deaths', 'Recovered', 'Active']]
      .sum()
      .query('Deaths > 50000')
      .sort_values('Deaths', ascending=False)
)

ax = top_countries.plot(kind='line', figsize=(14, 8), marker='o')
ax.set_title('Country-wise totals for countries with more than 50,000 deaths')
ax.set_xlabel('Country')
ax.set_ylabel('Number of cases')
ax.legend(title='Metric')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### Question 6

### Plot Province/State wise Deaths in USA

```python
import plotly.express as px
```

```python
covid_data= pd.read_csv('https://raw.githubusercontent.com/CSSEGISandData/COVID-19/master/csse_covid_19_data/csse_covid_19_daily_reports/01-09-2021.csv')
```

```python
covid_data.columns
```

```python
country = 'US'
state_deaths = (
    covid_data[covid_data['Country_Region'] == country]
      .groupby('Province_State', as_index=False)['Deaths']
      .sum()
)

fig = px.bar(
    state_deaths,
    x='Province_State',
    y='Deaths',
    title=f'{country} Province/State-wise Deaths',
    labels={'Province_State': 'Province/State', 'Deaths': 'Deaths'}
)
fig.update_layout(xaxis_tickangle=-45, yaxis_title='Deaths')
fig.show()
```

### Question 9

### Plot Worldwide Confirmed Cases over time

```python
import plotly.express as px
import plotly.io as pio
```

```python
from datetime import datetime

start_date = datetime(2021, 1, 1)
end_date = datetime(2021, 12, 31)
all_dates = pd.date_range(start=start_date, end=end_date, freq='D')

dfs_list = []
for date in all_dates:
    date_str = date.strftime('%m-%d-%Y')
    try:
        url = f'https://raw.githubusercontent.com/CSSEGISandData/COVID-19/master/csse_covid_19_data/csse_covid_19_daily_reports/{date_str}.csv'
        df_temp = pd.read_csv(url)
        df_temp['Date'] = date
        dfs_list.append(df_temp[['Date', 'Confirmed']].groupby('Date').sum().reset_index())
    except Exception:
        continue

worldwide_trend = pd.concat(dfs_list, ignore_index=True).sort_values('Date')

fig = px.line(
    worldwide_trend,
    x='Date',
    y='Confirmed',
    title='Worldwide Confirmed Cases Over Time (2021)',
    labels={'Date': 'Date', 'Confirmed': 'Confirmed Cases'},
)
fig.update_traces(line=dict(width=1))
fig.update_layout(xaxis_title='Date', yaxis_title='Confirmed Cases', hovermode='x unified')
fig.show()
```
